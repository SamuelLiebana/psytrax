# Compare built-in models

This notebook compares the built-in contrast GLM, DDM, and race-model likelihoods on the same first 500 trials from DAP009. The full script version lives at `examples/compare_models_DAP009.py`.

![DAP009 model evidence comparison](../_static/images/examples/model-evidence-comparison.png)

The comparison is reported as relative log evidence against the GLM baseline. Larger values are better, and the GLM is set to zero for reference.

## Load the three saved fits

Each fit was created with the same `psytrax.fit` call pattern but a different built-in likelihood and parameter initialisation.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

repo = Path.cwd().parents[1]
fit_dir = repo / "fits"

fit_files = {
    "GLM": fit_dir / "DAP009_logistic_N500_fit.npy",
    "DDM": fit_dir / "DAP009_ddm_N500_fit.npy",
    "Race": fit_dir / "DAP009_race_N500_fit.npy",
}

fits = {
    name: np.load(path, allow_pickle=True).item()
    for name, path in fit_files.items()
}

{name: fit["log_evidence"] for name, fit in fits.items()}
```

## Plot relative evidence

The GLM uses choices only, while the DDM and race model use choices and reaction times. This example is primarily a template for running the same comparison on your own dataset.

```python
baseline = fits["GLM"]["log_evidence"]
labels = list(fits)
delta = [fits[name]["log_evidence"] - baseline for name in labels]

fig, ax = plt.subplots(figsize=(7, 3.6))
colors = ["#2f7d67", "#4aa1bd", "#d48a2f"]
ax.barh(labels, delta, color=colors)
ax.axvline(0, color="#14313b", lw=1)
ax.set_xlabel("Δ log evidence vs. GLM (nats)")
ax.set_title("DAP009, first 500 trials")
ax.invert_yaxis()
fig.tight_layout()
```
